```python
import pandas as pd
```

```python
url = 'https://raw.githubusercontent.com/frontiertechinstitute/datasets/refs/heads/main/AI%20and%20Digital%20Media/Global_AI_Content_Impact_Dataset.csv'

df = pd.read_csv(url)
df.head()
```

```python
df.info()
```

```python
# count how many duplicates are in the dataset / drop duplicates
df.duplicated().sum()
df = df.drop_duplicates()
print(df.duplicated().sum())
```

```python
# count how many missing values are in the dataset : none
df.isna().sum()
```

```python
df.dtypes
```

```python
# Remove spaces
df['Country'] = df['Country'].str.strip()
print(df['Country'].unique())
```

```python
# Check if values are outside 0-100
out_of_range = df[
    (df['AI Adoption Rate (%)'] < 0) |
    (df['AI Adoption Rate (%)'] > 100)
]
print('AI Adoption Rate (%):', len(out_of_range))

out_of_range = df[
    (df['Job Loss Due to AI (%)'] < 0) |
    (df['Job Loss Due to AI (%)'] > 100)
]
print('Job Loss Due to AI (%):', len(out_of_range))

out_of_range = df[
    (df['Revenue Increase Due to AI (%)'] < 0) |
    (df['Revenue Increase Due to AI (%)'] > 100)
]
print('Revenue Increase Due to AI (%):', len(out_of_range))

out_of_range = df[
    (df['Human-AI Collaboration Rate (%)'] < 0) |
    (df['Human-AI Collaboration Rate (%)'] > 100)
]
print('Human-AI Collaboration Rate (%):', len(out_of_range))

out_of_range = df[
    (df['Consumer Trust in AI (%)'] < 0) |
    (df['Consumer Trust in AI (%)'] > 100)
]
print('Consumer Trust in AI (%):', len(out_of_range))

out_of_range = df[
    (df['Market Share of AI Companies (%)'] < 0) |
    (df['Market Share of AI Companies (%)'] > 100)
]
print('Market Share of AI Companies (%):', len(out_of_range))
```

```python
# check Year values are reasonable
print(df['Year'].min(), df['Year'].max())
```
