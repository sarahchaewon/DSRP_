```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
url = 'https://raw.githubusercontent.com/frontiertechinstitute/datasets/refs/heads/main/AI%20and%20Digital%20Media/Global_AI_Content_Impact_Dataset.csv'

df = pd.read_csv(url)
df.head()
```

```python
df.info()
```

```python
# no duplicates
df.duplicated().sum()
```

```python
# no missing values
df.isna().sum()
```

```python
# no need for standardization
for col in ['Country', 'Industry', 'Top AI Tools Used', 'Regulation Status'] :
  print(col, df[col].unique())
```

```python
# check for nonsensical values
year_out_of_range = df[(df["Year"]<2020)|(df["Year"]>2025)]
print(len(year_out_of_range))

percentage_columns = ["AI Adoption Rate (%)", "Job Loss Due to AI (%)", "Revenue Increase Due to AI (%)", "Human-AI Collaboration Rate (%)", "Consumer Trust in AI (%)", "Market Share of AI Companies (%)"]
for col in percentage_columns :
  percentage_out_of_range = df[(df[col]<0) | (df[col]>100)]
print(len(percentage_out_of_range))
```

```python
# Creating a scatterplot for relationship between Consumer Trust in AI and Revenue Increase Due to AI
x = df["Consumer Trust in AI (%)"]
y = df["Revenue Increase Due to AI (%)"]
plt.scatter(x,y,c='blue', marker='o')
plt.xlabel("Consumer Trust in AI (%)")
plt.ylabel("Revenue Increase Due to AI (%)")
plt.title("Relationship Between Consumer Trust in AI and Revenue Increase Due to AI")
```

```python
# Creating a scatterplot for relationship between Job Loss Due to AI and Revenue Increase Due to AI
x = df["Job Loss Due to AI (%)"]
y = df["Revenue Increase Due to AI (%)"]
plt.scatter(x, y, color='purple', marker='*')
plt.xlabel('Job Loss Due to AI (%)')
plt.ylabel('Revenue Increase Due to AI (%)')
plt.title('Job Loss vs Revenue Increase')
```

```python
# Creating a scatterplot for relationship between AI Adoption Rate and Job Loss Due to AI
x = df["AI Adoption Rate (%)"]
y = df["Job Loss Due to AI (%)"]
plt.scatter(x, y, color='red', marker='^')
plt.xlabel('AI Adoption Rate (%)')
plt.ylabel('Job Loss Due to AI (%)')
plt.title('AI Adoption Rate (%) vs Job Loss Due to AI (%)')
```
